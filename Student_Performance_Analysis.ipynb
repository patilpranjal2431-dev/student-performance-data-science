# Student Performance Data Science Mini Project

## Overview
This project analyzes a student-performance dataset to identify patterns associated with academic outcomes.

### Questions
1. How are students' total scores distributed?
2. How are grades distributed?
3. Is weekly self-study time associated with total score?
4. Is attendance associated with total score?
5. Is class participation associated with total score?
6. Which numerical factors have the strongest relationship with total score?

### Tools
- Python
- Pandas
- NumPy
- Matplotlib
- Seaborn

**Dataset:** `student_performance.csv`


```python
# 1. Import libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")
pd.set_option("display.max_columns", None)

print("Libraries imported successfully.")
```

```python
# 2. Load the dataset
df = pd.read_csv("student_performance.csv")

print(f"Dataset shape: {df.shape[0]:,} rows × {df.shape[1]} columns")
display(df.head())
```

```python
# 3. Understand the dataset
print("Column names:")
print(df.columns.tolist())

print("\nData types:")
print(df.dtypes)

print("\nMissing values:")
display(df.isnull().sum())

print(f"Duplicate rows: {df.duplicated().sum():,}")
```

```python
# 4. Descriptive statistics
display(df.describe())
```

## 5. Data quality and preparation

The dataset contains an identifier (`student_id`) and five student-performance variables.

For the analysis:
- `student_id` is treated as an identifier, not an analytical feature.
- Numerical columns are used for summary statistics and relationship analysis.
- `grade` is treated as a categorical outcome for distribution analysis.


```python
# 5. Prepare analytical columns
numeric_columns = [
    "weekly_self_study_hours",
    "attendance_percentage",
    "class_participation",
    "total_score"
]

categorical_column = "grade"

print("Numerical columns:")
print(numeric_columns)

print("\nGrade categories:")
print(sorted(df[categorical_column].dropna().unique()))
```

```python
# 6. Distribution of total scores
plt.figure(figsize=(9, 5))
sns.histplot(df["total_score"], bins=30, kde=True)
plt.title("Distribution of Total Scores")
plt.xlabel("Total Score")
plt.ylabel("Number of Students")
plt.tight_layout()
plt.show()
```

```python
# 7. Grade distribution
grade_counts = df["grade"].value_counts().sort_index()

plt.figure(figsize=(8, 5))
sns.barplot(x=grade_counts.index, y=grade_counts.values)
plt.title("Grade Distribution")
plt.xlabel("Grade")
plt.ylabel("Number of Students")
plt.tight_layout()
plt.show()

display(grade_counts.to_frame("Student Count"))
```

```python
# 8. Weekly self-study hours vs total score
plt.figure(figsize=(9, 6))
sns.scatterplot(
    data=df,
    x="weekly_self_study_hours",
    y="total_score",
    alpha=0.25
)
sns.regplot(
    data=df,
    x="weekly_self_study_hours",
    y="total_score",
    scatter=False
)
plt.title("Weekly Self-Study Hours vs Total Score")
plt.xlabel("Weekly Self-Study Hours")
plt.ylabel("Total Score")
plt.tight_layout()
plt.show()
```

```python
# 9. Attendance vs total score
plt.figure(figsize=(9, 6))
sns.scatterplot(
    data=df,
    x="attendance_percentage",
    y="total_score",
    alpha=0.25
)
sns.regplot(
    data=df,
    x="attendance_percentage",
    y="total_score",
    scatter=False
)
plt.title("Attendance vs Total Score")
plt.xlabel("Attendance (%)")
plt.ylabel("Total Score")
plt.tight_layout()
plt.show()
```

```python
# 10. Class participation vs total score
plt.figure(figsize=(9, 6))
sns.scatterplot(
    data=df,
    x="class_participation",
    y="total_score",
    alpha=0.25
)
sns.regplot(
    data=df,
    x="class_participation",
    y="total_score",
    scatter=False
)
plt.title("Class Participation vs Total Score")
plt.xlabel("Class Participation")
plt.ylabel("Total Score")
plt.tight_layout()
plt.show()
```

```python
# 11. Correlation matrix
correlation_matrix = df[numeric_columns].corr()

plt.figure(figsize=(8, 6))
sns.heatmap(
    correlation_matrix,
    annot=True,
    fmt=".2f",
    cmap="coolwarm",
    center=0
)
plt.title("Correlation Matrix")
plt.tight_layout()
plt.show()

display(correlation_matrix)
```

```python
# 12. Average performance by grade
grade_summary = (
    df.groupby("grade")["total_score"]
      .agg(["count", "mean", "median", "min", "max"])
      .sort_index()
)

display(grade_summary)
```

```python
# 13. Compare study, attendance and participation by grade
grade_feature_summary = (
    df.groupby("grade")[
        ["weekly_self_study_hours", "attendance_percentage", "class_participation", "total_score"]
    ]
    .mean()
    .sort_index()
)

display(grade_feature_summary.round(2))
```

```python
# 14. Identify the strongest numerical relationships with total score
correlations_with_score = (
    correlation_matrix["total_score"]
    .drop("total_score")
    .sort_values(key=lambda x: x.abs(), ascending=False)
)

print("Correlation with total score:")
display(correlations_with_score.to_frame("Correlation"))
```
